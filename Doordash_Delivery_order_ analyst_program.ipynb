part 5: 用户体验分析与漏斗模型
本部分将对用户评分进行分组与转化率分析，并通过SQL + pandas双版本实现漏斗分析模型（从浏览，下单，好评）。

目的：
---了解特定时间段内外卖好评率的占比
---分析用户在浏览，下单，评论各环节的流失情况

```python
df['RunDate'].value_counts().sort_index()
df['RunDate'].unique()
```

```python
#SQL评分体验分组统计
experience = pd.read_sql('''
select
    case
        when review_rating >=4.5 then '良好'
        when review_rating >=3.0 then '一般'
        else '糟糕'
    end as experience,
    count(*) as count
from doordash
group by experience
''', conn)
```

```python
#好评率分析(pandas)
df['is_good'] = df['review_rating'] >= 4.5
total_orders = df['review_rating'].count()
good_orders = df['is_good'].sum()
conversion_rate = good_orders / total_orders
print(f'好评率为: {conversion_rate:.2%}, ({good_orders}/{total_orders})')
```

```python
#SQL漏斗分析 + 转化率输出
sql_funnel = pd.read_sql('''
select
    count (*) as total_rows,
    count(case when url is not null then 1 end) as browsed,
    count(case when review_rating is not null then 1 end) as ordered,
    count(case when review_rating >= 4.5 then 1 end) as good_reviews
from doordash
''', conn)

#print(sql_funnel)

browsed = sql_funnel.at[0, 'browsed']
ordered = sql_funnel.at[0, 'ordered']
good_reviews = sql_funnel.at[0, 'good_reviews']

print(f'下单转化率:{ordered / browsed:.2%}')
print(f'好评转化率:{good_reviews / ordered:.2%}')
print(f'整体转化率:{good_reviews / browsed:.2%}')
```
